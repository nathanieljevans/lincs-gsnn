```python
import pandas as pd 
```

```python
prism1 = pd.read_csv('../../data/primary-screen-replicate-collapsed-logfold-change.csv').rename(columns={'Unnamed: 0': 'depmap_id'})
prism2 = pd.read_csv('../../data/secondary-screen-replicate-collapsed-logfold-change.csv').rename(columns={'Unnamed: 0': 'depmap_id'})
prism = prism1.merge(prism2, on='depmap_id', how='outer')
prism.head() 
```

```python
# need to map depmap id to cell iname 
info1 = pd.read_csv('../../data/DepMap-2019q1-celllines_v2.csv')[['DepMap_ID', 'CCLE_Name']].rename(columns={'DepMap_ID': 'depmap_id', 'CCLE_Name': 'ccle_name'})
info2 = pd.read_csv('../../data/cellinfo_beta.txt', sep='\t')[['ccle_name', 'cell_iname']]
info = pd.merge(info1, info2, on='ccle_name', how='inner')[['depmap_id', 'cell_iname']]
info.to_csv('../extdata/depmap2iname.csv', index=False)
info.head()  
```

```python
prism = prism.merge(info, on='depmap_id', how='inner')
prism.head() 
```

```python
prism = prism.set_index('cell_iname').drop(columns=['depmap_id']).stack().reset_index().rename(columns={'level_1': 'column_name', 0: 'logfold_change'})
prism.head() 
```

```python
meta1 = pd.read_csv('../../data/primary-screen-replicate-collapsed-treatment-info.csv')
meta2 = pd.read_csv('../../data/secondary-screen-replicate-collapsed-treatment-info.csv')
meta = pd.concat([meta1, meta2], axis=0)
meta = meta.assign(pert_id = [x.split('-00')[0] for x in meta['broad_id']])
meta = meta[['column_name', 'pert_id', 'name', 'dose', 'screen_id']]

prism = prism.merge(meta, on='column_name', how='inner') 
prism.head() 
```
